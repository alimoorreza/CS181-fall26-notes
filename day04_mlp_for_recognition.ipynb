## **Question 0.1: Create an MLP with four layers connected one after the other, following these specifications:**


> The first layer has 2 input nodes and 4 output nodes.

> The second layer takes 4 input nodes and outputs 8 nodes.

> The third layer connects 8 input nodes to 16 output nodes.

> The final layer connects 16 input nodes to 2 output nodes.

**Each layer, except the last one, should use the *ReLU()* activation function for its output nodes.**

```python
# import torch library
import torch
import numpy as np
import torch.nn as nn


# construction of a linear layer
torch.manual_seed(2)                    # for reproducibility
my_simple_mlp = nn.Sequential(
                          nn.Linear(2, 4),
                          nn.Linear(4, 8),
                          nn.Linear(8, 16),
                          nn.Linear(16, 2)
)

print(my_simple_mlp)
print("weight[0]:\n", my_simple_mlp[0].weight.data)
print("bias[0]:\n", my_simple_mlp[0].bias.data)
```

##**Question 0.2: Display the weights the first, second, and third layer of your MLP.**

```python
# the weights of the first layer
print(f'Weights of the first layer: \n{my_simple_mlp[0].weight.data}')

# the biases of the first layer
print(f'Biases of the first layer:  \n{my_simple_mlp[0].bias.data}\n\n')

# total number of parameters in the first layer
weight_matrix_param_first_layer     = my_simple_mlp[0].weight.data.shape[0]*my_simple_mlp[0].weight.data.shape[1]
bias_param_first_layer              = my_simple_mlp[0].bias.data.shape[0]
total_param_first_layer             = weight_matrix_param_first_layer + bias_param_first_layer
print(f"Total number of parameters (first layer):", total_param_first_layer)
```

```python
# the weights of the second layer
print(f'Weights of the second layer: \n{my_simple_mlp[1].weight.data}')

# the biases of the second layer
print(f'Biases of the second layer:  \n{my_simple_mlp[1].bias.data}\n\n')

# total number of parameters in the second layer
weight_matrix_param_second_layer    = my_simple_mlp[1].weight.data.shape[0]*my_simple_mlp[2].weight.data.shape[1]
bias_param_second_layer             = my_simple_mlp[1].bias.data.shape[0]
total_param_second_layer            = weight_matrix_param_second_layer + bias_param_second_layer
print(f"Total number of parameters (second layer):", total_param_second_layer)
```

```python
# the weights of the third layer
print(f'Weights of the third layer: \n{my_simple_mlp[2].weight.data}')

# the biases of the third layer
print(f'Biases of the third layer:  \n{my_simple_mlp[2].bias.data}\n\n')

# total number of parameters in the third layer
weight_matrix_param_third_layer    = my_simple_mlp[2].weight.data.shape[0]*my_simple_mlp[2].weight.data.shape[1]
bias_param_third_layer             = my_simple_mlp[2].bias.data.shape[0]
total_param_third_layer            = weight_matrix_param_third_layer + bias_param_third_layer
print(f"Total number of parameters (third) layer):", total_param_third_layer)
```

```python
# Print the weights of the final layer
print(f'Weights of the final layer: \n{my_simple_mlp[3].weight.data}')

# Print the biases of the final layer
print(f'Biases of the final layer:  \n{my_simple_mlp[3].bias.data}\n\n')

# total number of parameters in the final layer
weight_matrix_param_final_layer    = my_simple_mlp[3].weight.data.shape[0]*my_simple_mlp[3].weight.data.shape[1]
bias_param_final_layer             = my_simple_mlp[3].bias.data.shape[0]
total_param_final_layer            = weight_matrix_param_final_layer + bias_param_final_layer
print(f"Total number of parameters (final layer):", total_param_final_layer)
```

```python
# total number of parameters in the entire MLP

total_parameters =   total_param_first_layer + \
                            total_param_second_layer + \
                            total_param_third_layer + \
                            total_param_final_layer

print(f"Total number of parameters (the full network MLP):", total_parameters)
print(f"If our computer uses a standard single-precision floating-point number for each parameter, then each parameter occupies 4 bytes (32 bits) of memory.")
print(f"Hence, our computer needs to reserve 4*{total_parameters}={4*total_parameters} bytes of memory for this network.")
print(f"One kilobytes=1000 bytes")
print(f"One megabytes=1000 kilobytes")
print(f"One megabytes=1000*1000 bytes")
print(f"Hence, our computer needs to reserve 4*{total_parameters}/1000000={(4*total_parameters)/1000000} megabytes of memory for this network.")
```
