## **Let's build a 2D convolution layer**
[nn.Conv2d()](https://pytorch.org/docs/stable/generated/torch.nn.Conv2d.html)
  - applies a 2D convolution over an input volume of $(C_{in}​,H_{in},W_{in})$ and produces an output volume of $(C_{out}​,H_{out},W_{out})$   between two adjacent layers. For example, to produce the toy example below, we have $(C_{in}=8,H_{in}=6,W_{in}=7)$, which produces an output volume of $(C_{out}=8,H_{out}=4,W_{out}=5)$. Each filter in the middle has a __filter_size__ of $F=3$.
  - to create this, you need to provide the followings:
    - __channel_dimension_of_input_layer__ i.e., $C_{in}$
    - __channel_dimension_of_output_layer__ i.e., $C_{out}$
    - __filter_size__ i.e., $F$

  - the other two optional parameters are __stride__: $S=1$ and __padding__: $P=0$, with default values as shown. As discussed in class, PyTorch will calculate internally the sizes of output volume $H_{out}$ and $W_{out}$ from the above mentioned parameter values


> For example, let's create a 2D convolution filter bank like the image below by specify the following:
>> input volume channels size is $C_{in}=8$

>> output volume channel size is  $C_{out}=8$

>> each filter has a size of (3x3), so $F=3$

>> default stride size is 1

>> default padding size is 0

Hence, you provide the following PyTorch code to create this Conv2d layer:
>> cnn1 = nn.Conv2d( 8, 8, 3, 1, 0)

```python
# import torch library
import torch
import numpy as np
import torch.nn as nn

conv_layer = nn.Conv2d( 8, 8, 3, 1, 0)
```

### **Create a random input volume of the given size.**

> More precisely, the input has 8 channels, a height of 6, and a width of 7. In other words, $(C_{in}=8, H_{in}=6, W_{in}=7)$.

```python
input_volume_tensor = torch.randn((1, 8, 6, 7))
print(f"input volume shape:", input_volume_tensor.shape)    # [HOW_MANY_TENSORS, CHANNEL_SIZE, HEIGHT, WIDTH]
```

### **Perform a forward pass of the input volume through the newly created convolutional layer.**

> More precisely, the input has 8 channels, a height of 6, and a width of 7. In other words, $(C_{in}=8, H_{in}=6, W_{in}=7)$.

```python
output_volume_tensor = conv_layer(input_volume_tensor)
print(f"output volume shape:", output_volume_tensor.shape) # [HOW_MANY_TENSORS, CHANNEL_SIZE, HEIGHT, WIDTH]
```

###**Verify the shape of the output volume $(C_{out}​,H_{out},W_{out})$ by hand calculation**
There are 8 filters in this convolutional layer. Each filter has a size of (3x3), hence:

- $F=3$

You should use the default values of the other two optional parameters as follows (when they are not mentioned explicitly):
-  __stride__: $S=1$
- __padding__: $P=0$

Given the above parameters and our input volume of sizes $C_{in}$=8, $H_{in}$=6, $W_{in}$=6, we need to find/verify the following:

- $C_{out}$=?

- $H_{out}$=?

- $W_{out}$=?


Total number of convolutional filters used in this layer is $8$, hence $C_{out}$=8


To find the size of the output volume's height by hand calculation do the following:

> $H_{out}$ = ($H_{in}$ - $F$ + 2*$P$)/$S$ + 1

>> (6 - 3 + 2*0)/1 + 1

>> = 3/1 + 1

>> = 3 + 1

>> = 4

Hence output volume's height $H_{out}$ = 4

To find the size of the output volume's width by hand calculation do the following:

> $W_{out}$ = ($W_{in}$ - $F$ + 2*$P$)/$S$ + 1

>> (7 - 3 + 2*0)/1 + 1

>> = 4/1 + 1

>> = 4 + 1

>> = 5

Hence output volume's width $W_{out}$ = 5

## **Question 0.1: create a CNN with 2 conv2d layers and connect them in a sequence following the specifications below:**


The first convolution layer has:
  > its input volume has 1 channels. It can take any arbitrary volume with $(C_{in}=3, H_{in}=?, W_{in}=?)$. For example, you could create an input volume of $(C_{in}=3, H_{in}=50, W_{in}=50)$.

  > the output volume will be of 2 channels, ie, $(C_{out}=6$

  > each filter has a size of (3x3) with a __stride__ and a __padding__ as follows:
  - $F=3$
  - $S=2$
  - $P=0$


Following a similar convention, the second convolution layer has:
  > its input volume has 2 channels

  > the output volume will be of 4 channels

  > each filter has a size of (3x3) with a __stride__ and a __padding__ as follows:
  - $F=3$
  - $S=2$
  - $P=0$

The third convolution layer has:
  > its input volume has 4 channels

  > the output volume will be of 8 channels

  > each filter has a size of (3x3) with a __stride__ and a __padding__ as follows:
  - $F=3$
  - $S=2$
  - $P=0$



**After each convolution layer you should use the *ReLU()* activation function for its output**
<!--**Use a default stride size is 1 (don't need to change)**
**Use a default padding size is 0 (don't need to change)**-->
> [Reference: nn.Conv2d()](https://pytorch.org/docs/stable/generated/torch.nn.Conv2d.html)


```python
my_cnn = nn.Sequential(
            nn.Conv2d(1, 2, 3, 2, 0),
            nn.ReLU(),
            nn.Conv2d(2, 4, 3, 2, 0),
            nn.ReLU(),
            nn.Conv2d(4, 8, 3, 2, 0),
            nn.ReLU()
)
```

##**Question 0.2: Display the weights of these three convolution layers of your CNN.**

```python
# the weights of the first layer
#print(f'Weights of the first layer: \n{my_cnn[0].weight.data}')
print(f'size of the filter weights (first layer): \n{my_cnn[0].weight.data.shape}')

# the biases of the first layer
print(f'size of the filter biases (first layer): \n{my_cnn[0].bias.data.shape}\n\n')

# total number of parameters in the first layer
weight_matrix_param_first_layer     = my_cnn[0].weight.data.shape[0]*my_cnn[0].weight.data.shape[1]*my_cnn[0].weight.data.shape[2]*my_cnn[0].weight.data.shape[3]
bias_param_first_layer              = my_cnn[0].bias.data.shape[0]
total_param_first_layer             = weight_matrix_param_first_layer + bias_param_first_layer
print(f"Number of filter weight parameters (first layer):", weight_matrix_param_first_layer)
print(f"Number of filter bias parameters (first layer):", bias_param_first_layer)
print(f"-"*60)
print(f"Total number of parameters (first layer):", total_param_first_layer)
print(f"="*60)

# ------------------------------------------------------------------------------------------------
# the weights of the second layer
#print(f'Weights of the second layer: \n{my_cnn[1].weight.data}')
print(f'size of the filter weights (second layer): \n{my_cnn[2].weight.data.shape}')

# the biases of the first layer
print(f'size of the filter biases (second layer): \n{my_cnn[2].bias.data.shape}\n\n')

# total number of parameters in the second layer
weight_matrix_param_second_layer     = my_cnn[2].weight.data.shape[0]*my_cnn[2].weight.data.shape[1]*my_cnn[2].weight.data.shape[2]*my_cnn[2].weight.data.shape[3]
bias_param_second_layer              = my_cnn[2].bias.data.shape[0]
total_param_second_layer             = weight_matrix_param_second_layer + bias_param_second_layer
print(f"Number of filter weight parameters (second layer):", weight_matrix_param_second_layer)
print(f"Number of filter bias parameters (second layer):", bias_param_second_layer)
print(f"-"*60)
print(f"Total number of parameters (second layer):", total_param_second_layer)
print(f"="*60)

# ------------------------------------------------------------------------------------------------
# the weights of the third layer
#print(f'Weights of the third layer: \n{my_cnn[4].weight.data}')
print(f'size of the filter weights (third layer): \n{my_cnn[4].weight.data.shape}')

# the biases of the third layer
print(f'size of the filter biases (third layer): \n{my_cnn[4].bias.data.shape}\n\n')

# total number of parameters in the third layer
weight_matrix_param_third_layer     = my_cnn[4].weight.data.shape[0]*my_cnn[4].weight.data.shape[1]*my_cnn[4].weight.data.shape[2]*my_cnn[4].weight.data.shape[3]
bias_param_third_layer              = my_cnn[4].bias.data.shape[0]
total_param_third_layer             = weight_matrix_param_third_layer + bias_param_third_layer
print(f"Number of filter weight parameters (third layer):", weight_matrix_param_third_layer)
print(f"Number of filter bias parameters (third layer):", bias_param_third_layer)
print(f"-"*60)
print(f"Total number of parameters (third layer):", total_param_third_layer)
print(f"="*60)

# ------------------------------------------------------------------------------------------------
# Total size of your CNN (total number of parameters among all the convolution layers)

total_parameters =   total_param_first_layer + \
                            total_param_second_layer + \
                            total_param_third_layer

print(f"Total number of parameters (the full network CNN): {total_param_first_layer}+{total_param_second_layer}+{total_param_third_layer}={total_parameters}")
print(f"If our computer uses a standard single-precision floating-point number for each parameter, then each parameter occupies 4 bytes (32 bits) of memory.")
print(f"Hence, our computer needs to reserve 4*{total_parameters}={4*total_parameters} bytes of memory for this network.")
print(f"One kilobytes=1000 bytes")
print(f"One megabytes=1000 kilobytes")
print(f"One megabytes=1000*1000 bytes")
print(f"Hence, our computer needs to reserve 4*{total_parameters}/1000000={(4*total_parameters)/1000000} megabytes of memory for this network.")
```

Alternatively, you could use an object-oriented programming approach to create a class for this network. This would make it easier to inspect the output volumes at the intermediate layers.


```python
class MyCNN(nn.Module):

    def __init__(self):

        super(MyCNN, self).__init__()
        self.conv2d_layer1          = nn.Conv2d(1, 2, 3, 2, 0)
        self.conv2d_layer2          = nn.Conv2d(2, 4, 3, 2, 0)
        self.conv2d_layer3          = nn.Conv2d(4, 8, 3, 2, 0)
        self.relu_for_all_layers    = nn.ReLU()


    def forward(self, x):

        print("shape of input (     ): ", x.shape)
        x = self.conv2d_layer1(x)
        x = self.relu_for_all_layers(x)
        print("output shape (layer#1): ", x.shape)
        x = self.conv2d_layer2(x)
        x = self.relu_for_all_layers(x)
        print("output shape (layer#2): ", x.shape)
        x = self.conv2d_layer3(x)
        x = self.relu_for_all_layers(x)
        print("output shape (layer#3): ", x.shape)
        return x

my_cnn_standard = MyCNN()
```

```python
output                = my_cnn_standard(random_input_volume)
#print(f'output layer values: \n{output.data.numpy()}\n')
print(f'Size of the final output (last layer): \n{output.data.shape}\n')
```
