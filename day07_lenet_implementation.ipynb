## **Group Activity**: *Let’s try to mimic the CNN above by recreating a popular CNN called the **LeNet**. You should insert a $ReLU$ activation layer and a $max$-$pooling$ layer between each pair of adjacent Conv2D layers, as we did before, and connect the layers sequentially according to the specifications below:*



```python
# your code
# ...
# ...
# ...

import torch
import torch.nn as nn

class LeNetcs181(nn.Module):
    def __init__(self):
        super(LeNetcs181, self).__init__()
        self.conv_layer1        = nn.Conv2d(1, 6, 5, 1, 0)  # kernel=(5,5) and stride=(1,1)
        self.relu_layer1        = nn.ReLU()
        self.avgpooling_layer1  = nn.AvgPool2d(2, 2)        # kernel=(2,2) and stride=(2,2)

        self.conv_layer2        = nn.Conv2d(6, 16, 5, 1, 0) # kernel=(5,5) and stride=(1,1)
        self.relu_layer2        = nn.ReLU()
        self.avgpooling_layer2  = nn.AvgPool2d(2, 2)        # kernel=(2,2) and stride=(2,2)

        # transform the output volume into a long list of numbers (ie, volume --> vector transformation)

        # final mlp layers for classification
        self.linear_layer1      = nn.Linear(16*5*5, 84)
        self.linear_layer2      = nn.Linear(84, 10)

    def forward(self, x):

        # first layers of conv2d + relu + avgpooling2d
        output = self.conv_layer1(x)
        output = self.relu_layer1(output)
        print(f"shape after conv_layer1: {output.shape}")
        output = self.avgpooling_layer1(output)
        print(f"shape after avgpooling_layer1: {output.shape}")

        # second layers of conv2d + relu + avgpooling2d
        output = self.conv_layer2(output)
        output = self.relu_layer2(output)
        print(f"shape after conv_layer2: {output.shape}")
        output = self.avgpooling_layer2(output)
        print(f"shape after avgpooling_layer2: {output.shape}")

        # transform the output volume into a long list of numbers (ie, volume --> vector transformation)
        # use torch.flatten() operation
        print(f"before flattening the volume using torch.flatten() operation ...")
        output = torch.flatten(output, 1)
        print(f"after flattening the volume shape becomes {output.shape}")
        # two layers of MLP for classification
        output = self.linear_layer1(output)
        print(f"shape after linear_layer1: {output.shape}")
        output = self.linear_layer2(output)
        print(f"shape after linear_layer2: {output.shape}")

```
